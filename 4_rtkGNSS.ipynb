# RTK GNSS øvelse
## Indledende kode
Først skal vi importere de funktioner vi skal bruge til at beregne og visualisere vores observationer.

```python
from math import sqrt
from numpy import array, linspace, genfromtxt, mean, std
from matplotlib import pyplot as plt
import matplotlib
matplotlib.rcParams['axes.formatter.useoffset'] = False
matplotlib.rcParams['axes.formatter.limits'] = (-5, 8)
```

## Fikspunkter
Vi kan nu samle positionerne og spredningerne fra RTK observationerne vi har indsamlet under dern geodætiske sightseeing.

jhs?_true indeholder de kendte koordinater fra fikspunktsbeskrivelserne.

```python
jhs3_true = [382716.145, 7427154.541, 61.838]
jhs4_true = [383568.688, 7427024.514, 89.648]
jhs8_true = [383244.680, 7426448.034, 99.691]
jhs9_true = [383689.513, 7427185.099, 90.690]
```

## Observationer


```python
# henter data fra fil
rtkObs = genfromtxt("../../billhill_rtk_Gruppe1.txt", skip_header=1, delimiter=",", dtype="U20,3f8", usecols=(0,2,1,3))
# printer data fra fil
for (i, obs) in enumerate(rtkObs):
    print(i, obs)
```

## Visuel vurdering af fikspunktsobservationer observationer

Observationerne fra alle grupper kan nu plottes og samenlignes visuelt i forhold til den "sande position". 
Find alle de linier der matcher med punktnavnet, f.eks er JHS3 linie 1 i eksemplet ovenfor.
Erstat derefter første "?" med 1 i jhs3_obs_diff i første linie herunder.
Erstat herefter alle "?" med de linienumre der indeholder JHS3 observationer.

Gentag for JHS4, JHS8 og JHS9.

```python
# find difference mellem observation og angivne position af fikspunkter
jhs3_obs_diff = rtkObs[?][1] - jhs3_true  # Erstat det første ? med linienummeret på den første linie med JHS3
jhs4_obs_diff = rtkObs[?][1] - jhs4_true
jhs8_obs_diff = rtkObs[?][1] - jhs8_true
jhs9_obs_diff = rtkObs[?][1] - jhs9_true
```

```python
# visualiser differencen
plt.figure()
plt.scatter(jhs3_obs_diff[0], jhs3_obs_diff[1], label=f'JHS3')
plt.scatter(jhs4_obs_diff[0], jhs4_obs_diff[1], label=f'JHS4')
plt.scatter(jhs8_obs_diff[0], jhs8_obs_diff[1], label=f'JHS8')
plt.scatter(jhs9_obs_diff[0], jhs9_obs_diff[1], label=f'JHS9')

plt.xlabel('Easting')
plt.ylabel('Northing')
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()
```

## Beregn middelværdi og spredning for RTK observationerne
Se på jeres Jupyter notebook fra KodeGNSS øvelsen og beregn gennemsnitskoordinater og spredninger på jeres RTK observationer.

```python
jhs3_obs = array((rtkObs[1][1], rtkObs[2][1], rtkObs[3][1], rtkObs[0][1]))
jhs3_e_mean = mean(jhs3_obs[:, 0])
jhs3_e_std = std(jhs3_obs[:, 0], ddof=1)
jhs3_n_mean = mean(jhs3_obs[:, 1])
jhs3_n_std = std(jhs3_obs[:, 1], ddof=1)
jhs3_h_mean = mean(jhs3_obs[:, 2])
jhs3_h_std = std(jhs3_obs[:, 2], ddof=1)

print(f"JHS3\nE: {jhs3_e_mean:.3f} m +/- {jhs3_e_std:.3f} m\n" +
      f"N: {jhs3_n_mean:.3f} m +/- {jhs3_e_std:.3f} m\n" + 
      f"H: {jhs3_h_mean:.3f} m +/- {jhs3_h_std:.3f} m\n")
```
